```
Algorithm: BACKPROP(network, input, target)

begin
    // Forward pass
    for each layer l do
        z[l] ← W[l] × a[l-1] + b[l]
        a[l] ← σ(z[l])
    
    // Compute output error
    δ[L] ← (a[L] - target) ⊙ σ'(z[L])
    
    // Backward pass
    for l = L-1 down to 1 do
        δ[l] ← (W[l+1]^T × δ[l+1]) ⊙ σ'(z[l])
    
    // Update weights
    for each layer l do
        W[l] ← W[l] - α × δ[l] × a[l-1]^T
        b[l] ← b[l] - α × δ[l]
end
```

**Loss Function:**

$$L = \frac{1}{2}\sum_i (y_i - \hat{y}_i)^2$$

## 7.2 Implementation

See: [ch07_neural_implementation.ipynb](ch07_neural_implementation.ipynb)

Includes:
- Neural network from scratch
- Backpropagation
- Activation functions
- Optimization algorithms
- Real-world: MNIST digit recognition
