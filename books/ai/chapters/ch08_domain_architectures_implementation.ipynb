```python
import numpy as np

print('Imports complete')
```

```python
def conv2d(image, kernel):
    i_h, i_w = image.shape
    k_h, k_w = kernel.shape
    o_h, o_w = i_h - k_h + 1, i_w - k_w + 1
    output = np.zeros((o_h, o_w))
    
    for i in range(o_h):
        for j in range(o_w):
            output[i,j] = np.sum(image[i:i+k_h, j:j+k_w] * kernel)
    
    return output

def max_pool(image, pool_size=2):
    h, w = image.shape
    o_h, o_w = h // pool_size, w // pool_size
    output = np.zeros((o_h, o_w))
    
    for i in range(o_h):
        for j in range(o_w):
            output[i,j] = np.max(image[i*pool_size:(i+1)*pool_size, 
                                       j*pool_size:(j+1)*pool_size])
    
    return output

print('CNN operations defined')
```

```python
def attention(Q, K, V):
    d_k = Q.shape[-1]
    scores = np.matmul(Q, K.T) / np.sqrt(d_k)
    weights = np.exp(scores) / np.sum(np.exp(scores), axis=-1, keepdims=True)
    return np.matmul(weights, V)

print('Attention mechanism implemented')
```

```python
print('=== Attention Example ===')
Q = np.random.randn(3, 4)
K = np.random.randn(3, 4)
V = np.random.randn(3, 4)
result = attention(Q, K, V)
print(f'Output shape: {result.shape}')
```
