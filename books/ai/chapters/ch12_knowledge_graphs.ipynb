## 12.1 Introduction

Knowledge graphs represent one of the most significant bridges between **deductive reasoning** and **inductive learning** in artificial intelligence. While the term was popularized by Google in 2012 for search enhancement, knowledge graphs have existed since the early years of AI in the form of **semantic networks**.

### Historical Context

**Semantic networks** were graphical representations of relationships among entities, where:
- **Nodes** correspond to concepts
- **Edges** correspond to relationships among concepts

Early applications included natural language processing tasks like word-sense disambiguation. Modern vocabulary representations like **WordNet** are direct descendants of these early semantic networks.

### What is a Knowledge Graph?

A comprehensive definition (Paulheim 2017):

> A knowledge graph:
> 1. Mainly describes real-world entities and their interrelations, organized in a graph
> 2. Defines possible classes and relations of entities in a schema
> 3. Allows for potentially interrelating arbitrary entities
> 4. Covers various topical domains

### Knowledge Graphs vs. First-Order Logic

Knowledge graphs are closely related to first-order logic knowledge bases:
- **First-order logic**: Can have n-ary relations (predicates with any number of arguments)
- **Knowledge graphs**: Focus on binary relationships (edges between two nodes)

This simplification enables powerful machine learning techniques while maintaining expressiveness.

### Components

1. **TBox (Terminological Box)**
   - Ontology: Hierarchical taxonomy of concepts
   - Schema: Blueprint of the database structure
   - Concept nodes arranged as directed acyclic graphs

2. **ABox (Assertion Box)**
   - Instances of entities
   - Relationships among instances
   - Typically orders of magnitude larger than TBox

3. **RDF Triples**
   - (Subject, Predicate, Object)
   - Example: (Tom_Hanks, acted_in, Saving_Private_Ryan)

## 12.3 Real-World Knowledge Graphs

### 12.3.1 WordNet

**WordNet** is a lexical database organizing English words into cognitive synonym sets (synsets).

**Statistics:**
- ~117,000 synsets
- Covers nouns, verbs, adjectives, adverbs

**Relationship Types:**

1. **Synonymy**: Words with similar meanings in the same synset
2. **Hypernymy/Hyponymy**: General-to-specific relationships
   - Example: furniture → bed → bunkbed
3. **Meronymy**: Part-whole relationships
   - Example: chair → leg (part-of)
4. **Antonymy**: Opposite meanings (for adjectives)
   - Example: good ↔ bad
5. **Troponymy**: Verb hierarchies by specificity
   - Example: communicate → talk → whisper
6. **Entailment**: Event relationships
   - Example: buy → pay, succeed → try

**Applications:**
- Word sense disambiguation
- Semantic similarity computation
- Feature engineering for NLP tasks

### 12.3.2 YAGO (Yet Another Great Ontology)

**Source**: Max Planck Institute
**Statistics** (2020):
- 10 million entities
- 120 million facts

**Key Features:**
1. **Temporal dimension**: Relations tagged with dates
2. **Spatial integration**: Geographic information
3. **Confidence scores**: ~95% accuracy (manually validated)
4. **Multi-lingual**: Built from Wikipedia in multiple languages

**Example YAGO Structure:**
```
Max_Planck
  - type → Physicist
  - type → Nobel_Prize_Winner
  - born_in → Kiel [1858-04-23]
  - awarded → Nobel_Prize [1918]

Physicist → subclass_of → Scientist
Scientist → subclass_of → Person
```

## 12.4 Knowledge Graph Construction

### 12.4.1 Construction Methodologies

| Method | Schema | Examples | Pros/Cons |
|--------|--------|----------|-----------|
| **Curated** | Yes | WordNet, CYC, UMLS | High quality, doesn't scale |

### 12.4.2 Information Extraction Pipeline

Two critical steps for automated construction:

#### **Step 1: Named Entity Recognition (NER)**

Identify entities in unstructured text.

**Example Text:**
> "Bill Clinton lives in New York at a location a few miles from an IBM building."

**Extracted Entities:**
- Bill_Clinton: PERSON
- New_York: LOCATION
- IBM: ORGANIZATION

```
Algorithm: NAMED-ENTITY-RECOGNITION(text)

Input: Unstructured text string
Output: List of (entity, type) pairs

begin
    tokens ← tokenize(text)
    entities ← []
    
    for each sequence s in tokens do
        features ← extract_features(s)
        // Features: capitalization, position, context
        
        entity_type ← classifier.predict(features)
        // Trained on labeled examples
        
        if entity_type != NONE then
            entities.append((s, entity_type))
    
    return entities
end
```

**Common Methods:**
- Rule-based systems
- Hidden Markov Models (HMMs)
- Conditional Random Fields (CRFs)
- Deep learning: BiLSTM-CRF, BERT

#### **Step 2: Relation Extraction**

Extract relationships between identified entities.

**Example:**
> "Bill Clinton and his wife, Hillary Clinton, relocated to New York."

**Extracted Relations:**
- (Bill_Clinton, spouse_of, Hillary_Clinton)
- (Bill_Clinton, lives_in, New_York)

```
Algorithm: RELATION-EXTRACTION(text, entities)

Input: Text and list of entities
Output: List of (subject, predicate, object) triples

begin
    relations ← []
    
    for each entity pair (e1, e2) in entities do
        context ← extract_context(text, e1, e2)
        // Words between and around entities
        
        features ← [
            entity_types(e1, e2),
            dependency_path(e1, e2),
            keywords_in_context
        ]
        
        relation_type ← classifier.predict(features)
        
        if relation_type != NONE then
            relations.append((e1, relation_type, e2))
    
    return relations
end
```

### 12.4.3 Handling Incompleteness

Knowledge graphs are inherently incomplete due to:
- Manual construction errors
- Extraction errors from text
- Missing data in source documents

**Solution: Shared Matrix Factorization**

For a knowledge graph with $n$ entities and $m$ relation types, create matrices $\mathbf{D}_1, \ldots, \mathbf{D}_m$ where $\mathbf{D}_i$ is the $n \times n$ adjacency matrix for relation type $i$.

**Factorization:**

$$\mathbf{D}_i \approx \mathbf{U}_i \mathbf{V}^T \quad \text{for } i = 1, \ldots, m$$

Where:
- $\mathbf{V}$ is the $n \times k$ **shared factor matrix** (common across all relations)
- $\mathbf{U}_i$ is the $n \times k$ relation-specific factor
- $k$ is the embedding dimension

**Optimization Objective:**

$$J = \|\mathbf{D}_1 - \mathbf{U}_1\mathbf{V}^T\|_F^2 + \sum_{i=2}^m \lambda_i \|\mathbf{D}_i - \mathbf{U}_i\mathbf{V}^T\|_F^2$$

**Gradient Updates:**

$$\mathbf{U}_1 \leftarrow \mathbf{U}_1 - \eta \frac{\partial J}{\partial \mathbf{U}_1} = \mathbf{U}_1 - \eta (\mathbf{D}_1 - \mathbf{U}_1\mathbf{V}^T)\mathbf{V}$$

$$\mathbf{U}_i \leftarrow \mathbf{U}_i - \eta \lambda_i (\mathbf{D}_i - \mathbf{U}_i\mathbf{V}^T)\mathbf{V} \quad \text{for } i \geq 2$$

$$\mathbf{V} \leftarrow \mathbf{V} - \eta \left[ (\mathbf{D}_1 - \mathbf{U}_1\mathbf{V}^T)^T\mathbf{U}_1 + \sum_{i=2}^m \lambda_i (\mathbf{D}_i - \mathbf{U}_i\mathbf{V}^T)^T\mathbf{U}_i \right]$$

**Benefits:**
- Reconstructs missing edges
- Creates unified entity embeddings
- Enables multi-relational learning

```
Algorithm: SHARED-MATRIX-FACTORIZATION(D1, ..., Dm, k)

Input: Relation matrices D1, ..., Dm; embedding dimension k
Output: Factor matrices U1, ..., Um and shared V

begin
    n ← number of entities
    
    // Initialize randomly
    V ← random(n, k)
    for i = 1 to m do
        Ui ← random(n, k)
    
    // Gradient descent
    for iter = 1 to max_iterations do
        // Update U1
        U1 ← U1 - η * (D1 - U1*V^T) * V
        
        // Update Ui for i ≥ 2
        for i = 2 to m do
            Ui ← Ui - η * λi * (Di - Ui*V^T) * V
        
        // Update shared V
        grad_V ← (D1 - U1*V^T)^T * U1
        for i = 2 to m do
            grad_V ← grad_V + λi * (Di - Ui*V^T)^T * Ui
        V ← V - η * grad_V
        
        // Check convergence
        if change < tolerance then
            break
    
    return U1, ..., Um, V
end
```

## 12.5 Applications of Knowledge Graphs

### 12.5.1 Search and Question Answering

**Traditional Approach:**
1. Parse natural language query
2. Identify entities and relationships
3. Translate to structured query (SPARQL)
4. Execute on knowledge graph

**Example Query Translation:**

**Natural Language:**
> "Who played Obi-Wan Kenobi in Star Wars?"

**SPARQL Translation:**
```sparql
SELECT ?actor WHERE {
  ?actor acted_in ?movie .
  ?movie has_title "Star Wars" .
  ?role has_name "Obi-Wan Kenobi" .
  ?actor played_role ?role .
}
```

**Result:** Ewan McGregor

**Modern Approach: Sequence-to-Sequence Learning**

Train neural models to translate NL → SPARQL:
1. **Training Data**: (NL query, SPARQL query) pairs
2. **Model**: Transformer-based encoder-decoder
3. **Refinement**: User feedback (clicks) as implicit labels

### 12.5.2 Clustering Knowledge Graphs

**Goal**: Discover similar entities and create summaries

**Method**: Use shared embeddings from matrix factorization

```
Algorithm: KG-CLUSTERING(U1, ..., Um, V, num_clusters)

Input: Factor matrices from shared factorization
Output: Cluster assignments for entities

begin
    // Create unified representation
    n ← number of entities
    k ← embedding dimension
    
    representations ← []
    for i = 1 to n do
        // Concatenate entity's representation across all relations
        rep_i ← [U1[i], U2[i], ..., Um[i], V[i]]
        representations.append(rep_i)
    
    // Apply standard clustering
    clusters ← k-means(representations, num_clusters)
    
    return clusters
end
```

**Applications:**
- Movie clustering: Find similar movies, actors, directors
- Author clustering: Discover research communities
- Product clustering: Recommend similar items

### 12.5.3 Entity Classification

**Problem**: Predict missing entity types or properties

**Example**: Given a new person node connected to movies, determine if they are an actor or director.

**Approach**:
1. Create feature vectors from entity embeddings
2. Use neighboring node types and relation types
3. Train supervised classifier on labeled entities

```
Algorithm: ENTITY-CLASSIFICATION(kg, entity, embeddings)

Input: Knowledge graph, target entity, learned embeddings
Output: Predicted entity type

begin
    // Extract features
    features ← []
    
    // 1. Entity's own embedding
    features.append(embeddings[entity])
    
    // 2. Aggregate neighbor embeddings
    for each neighbor n of entity do
        relation ← edge_type(entity, n)
        features.append([
            embeddings[n],
            relation_embedding[relation]
        ])
    
    // 3. Graph structural features
    features.append([
        degree(entity),
        num_relations_by_type
    ])
    
    // Classify using trained model
    entity_type ← classifier.predict(features)
    
    return entity_type
end
```

### 12.5.4 Link Prediction

**Problem**: Predict missing edges in the knowledge graph

**Method 1: Matrix Completion**

After factorization $\mathbf{D}_i \approx \mathbf{U}_i\mathbf{V}^T$:
- Examine entries in $\mathbf{U}_i\mathbf{V}^T$
- High values indicate likely missing edges
- Rank all potential edges and report top-k

```
Algorithm: LINK-PREDICTION(U, V, relation_type, k)

Input: Factor matrices, relation type, number of predictions k
Output: Top-k predicted links

begin
    n ← number of entities
    i ← index of relation_type
    
    // Reconstruct matrix
    D_pred ← Ui * V^T
    
    // Collect scores for all entity pairs
    scores ← []
    for p = 1 to n do
        for q = 1 to n do
            if D_pred[p,q] > threshold and original Di[p,q] == 0 then
                // High score but edge doesn't exist
                scores.append((p, q, D_pred[p,q]))
    
    // Sort and return top-k
    scores.sort(by=score, descending=True)
    return scores[1:k]
end
```

**Method 2: TransE Embeddings**

Learn embeddings where: $\mathbf{h} + \mathbf{r} \approx \mathbf{t}$

- $\mathbf{h}$: Head entity embedding
- $\mathbf{r}$: Relation embedding
- $\mathbf{t}$: Tail entity embedding

**Training Loss:**

$$L = \sum_{(h,r,t) \in S} \sum_{(h',r,t') \in S'} [\gamma + d(\mathbf{h}+\mathbf{r}, \mathbf{t}) - d(\mathbf{h}'+\mathbf{r}, \mathbf{t}')]_+$$

Where:
- $S$: Positive triples (exist in KG)
- $S'$: Negative samples (corrupted triples)
- $d(\cdot, \cdot)$: Distance function (L1 or L2)
- $\gamma$: Margin hyperparameter
- $[x]_+ = \max(0, x)$: Hinge loss

```
Algorithm: TRANSE-TRAINING(triples, k, γ, epochs)

Input: RDF triples, embedding dimension k, margin γ
Output: Entity and relation embeddings

begin
    // Initialize embeddings
    entity_emb ← random(num_entities, k)
    relation_emb ← random(num_relations, k)
    
    // Normalize to unit sphere
    entity_emb ← normalize(entity_emb)
    
    for epoch = 1 to epochs do
        for each (h, r, t) in triples do
            // Positive score
            pos_score ← ||entity_emb[h] + relation_emb[r] - entity_emb[t]||
            
            // Generate negative sample
            (h', r, t') ← corrupt_triple(h, r, t)
            neg_score ← ||entity_emb[h'] + relation_emb[r] - entity_emb[t']||
            
            // Compute loss
            loss ← max(0, γ + pos_score - neg_score)
            
            if loss > 0 then
                // Update embeddings via SGD
                update_embeddings(h, r, t, h', t', loss)
            
            // Re-normalize entity embeddings
            entity_emb ← normalize(entity_emb)
    
    return entity_emb, relation_emb
end

function corrupt_triple(h, r, t):
    if random() < 0.5 then
        h' ← random_entity()  // Corrupt head
        return (h', r, t)
    else
        t' ← random_entity()  // Corrupt tail
        return (h, r, t')
```

### 12.5.5 Recommender Systems

**Integration with Ratings Matrix**

Given:
- $\mathbf{R}$: $u \times n$ ratings matrix (users × items)
- $\mathbf{D}_1, \ldots, \mathbf{D}_m$: $n \times n$ relation matrices

**Joint Factorization:**

$$\mathbf{R} \approx \mathbf{M}\mathbf{V}^T$$
$$\mathbf{D}_i \approx \mathbf{U}_i\mathbf{V}^T$$

Where $\mathbf{V}$ is shared between ratings and relations.

**Objective Function:**

$$J = \|\mathbf{R} - \mathbf{M}\mathbf{V}^T\|_F^2 + \sum_{i=1}^m \lambda_i \|\mathbf{D}_i - \mathbf{U}_i\mathbf{V}^T\|_F^2$$

**Benefits:**
- Cold start: Use graph structure for new items
- Explainability: Recommendations based on item relationships
- Improved accuracy: Multi-relational context

## 12.6 SPARQL Query Language

**SPARQL** (SPARQL Protocol and RDF Query Language) is the standard query language for RDF-based knowledge graphs.

### Basic SPARQL Syntax

**Example 1: Find all actors in Star Wars**

```sparql
PREFIX movie: <http://example.org/movie/>

SELECT ?actor
WHERE {
  ?movie movie:title "Star Wars" .
  ?actor movie:acted_in ?movie .
}
```

**Example 2: Find movies directed by Steven Spielberg with awards**

```sparql
PREFIX movie: <http://example.org/movie/>
PREFIX award: <http://example.org/award/>

SELECT ?movie ?award
WHERE {
  movie:Steven_Spielberg movie:directed ?movie .
  ?movie award:received ?award .
}
```

**Example 3: Complex query with filtering**

```sparql
SELECT ?actor ?movie
WHERE {
  ?actor movie:acted_in ?movie .
  ?movie movie:director movie:Steven_Spielberg .
  ?movie award:received ?award .
  FILTER (?award = award:Best_Picture)
}
ORDER BY ?actor
```

### SPARQL Features

1. **OPTIONAL**: Include optional patterns
2. **FILTER**: Apply conditions
3. **UNION**: Combine alternative patterns
4. **ORDER BY**: Sort results
5. **LIMIT**: Restrict number of results
6. **COUNT/SUM/AVG**: Aggregation functions

## 12.7 Summary

Knowledge graphs bridge deductive reasoning and inductive learning in AI:

**Key Concepts:**
1. **Structure**: RDF triples organized into TBox (ontology) and ABox (instances)
2. **Real-world KGs**: WordNet, YAGO, DBpedia, Freebase, Wikidata
3. **Construction**: Curated, collaborative, and automated methods
4. **Extraction**: Named entity recognition and relation extraction
5. **Completion**: Shared matrix factorization and TransE embeddings

**Applications:**
- **Search**: Entity-based queries and question answering
- **Clustering**: Group similar entities across relations
- **Classification**: Predict entity types from graph context
- **Link Prediction**: Discover missing relationships
- **Recommendations**: Enhance collaborative filtering with graph structure

**Modern Trends:**
- Machine learning methods now dominate over pure reasoning
- Deep learning for entity/relation extraction
- Graph neural networks for embeddings
- Neuro-symbolic integration (Chapter 13)

**Trade-offs:**
- **Curated KGs**: High quality but limited scale
- **Automated KGs**: Large scale but with errors
- **Open vs. Domain-specific**: Breadth vs. depth

Knowledge graphs represent structured knowledge that combines the interpretability of symbolic AI with the power of statistical machine learning.
