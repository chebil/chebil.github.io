```
Algorithm: VARIABLE-ELIMINATION(query, evidence, bn)

begin
    factors ← get all CPTs from bn
    
    // Incorporate evidence
    for each variable e in evidence do
        factors ← restrict factors to e = evidence[e]
    
    // Eliminate hidden variables
    hidden ← variables - {query} - evidence
    for each variable h in hidden do
        // Sum out h
        relevant ← factors containing h
        product ← multiply all relevant factors
        new_factor ← sum over h in product
        factors ← factors - relevant + new_factor
    
    // Multiply remaining factors
    result ← multiply all factors
    
    // Normalize
    return normalize(result)
end
```

## 11.3 Sampling Methods

### Gibbs Sampling

```
Algorithm: GIBBS-SAMPLING(bn, query, evidence, N)

begin
    // Initialize state randomly
    state ← random assignment consistent with evidence
    counts ← zeros
    
    for i = 1 to N do
        for each non-evidence variable X do
            // Sample X given Markov blanket
            state[X] ← sample from P(X | MarkovBlanket(X))
        
        counts[state[query]] += 1
    
    return normalize(counts)
end
```

## 11.4 Implementation

See: [ch11_pgm_implementation.ipynb](ch11_pgm_implementation.ipynb)

Includes:
- Bayesian network structure
- Variable elimination
- Gibbs sampling
- Belief propagation
- Real-world: Medical diagnosis system
