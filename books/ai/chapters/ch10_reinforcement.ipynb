```
Algorithm: Q-LEARNING(episodes, α, γ, ε)

begin
    Initialize Q(s,a) arbitrarily
    
    for each episode do
        s ← initial state
        
        repeat
            // ε-greedy action selection
            if random() < ε then
                a ← random action
            else
                a ← argmax_a Q(s,a)
            
            // Take action, observe reward and next state
            s', r ← environment.step(a)
            
            // Q-update
            Q(s,a) ← Q(s,a) + α[r + γ max_a' Q(s',a') - Q(s,a)]
            
            s ← s'
        
        until s is terminal
    
    return Q
end
```

**Update Rule:**

$$Q(s,a) \leftarrow Q(s,a) + \alpha[r + \gamma \max_{a'} Q(s',a') - Q(s,a)]$$

## 10.2 Implementation

See: [ch10_reinforcement_implementation.ipynb](ch10_reinforcement_implementation.ipynb)

Includes:
- Q-Learning
- SARSA
- Policy gradient
- Value iteration
- Real-world: GridWorld, CartPole
