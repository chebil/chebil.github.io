## 10.1 Introduction

**Reinforcement Learning (RL)** is learning through interaction with an environment to maximize cumulative reward.

### RL vs Other Learning Paradigms

| Paradigm | Data | Feedback | Goal |
|----------|------|----------|------|
| Supervised | Labeled $(x,y)$ | Correct answer | Minimize error |
| Unsupervised | Unlabeled $x$ | None | Find structure |
| Reinforcement | State-action sequences | Delayed reward | Maximize cumulative reward |

### Key Characteristics

1. **Trial-and-error**: Agent learns by exploring
2. **Delayed rewards**: Actions have long-term consequences
3. **Sequential decisions**: Current choice affects future options
4. **Exploration vs Exploitation**: Balance trying new actions vs using known good ones

### Agent-Environment Interface

```
     Agent
    ┌────────┐
    │        │
    │ Policy │
    │        │
    └───┬──┬──┘
       │  │
   a_t │  │ r_t, s_t
       │  │
    ┌──┴──┴──┐
    │        │
    │  Env  │
    │        │
    └────────┘
```

At each time $t$:
- Agent observes **state** $s_t$
- Agent selects **action** $a_t$
- Environment returns **reward** $r_t$ and next state $s_{t+1}$

### Applications

- **Game playing**: Chess, Go, Atari, StarCraft
- **Robotics**: Manipulation, locomotion, navigation
- **Autonomous vehicles**: Self-driving cars
- **Resource management**: Data center cooling, traffic control
- **Finance**: Trading, portfolio optimization
- **Healthcare**: Treatment optimization

## 10.2 Markov Decision Processes (MDPs)

**MDP** provides mathematical framework for sequential decision-making.

### 10.2.1 MDP Definition

An MDP is a tuple $(\mathcal{S}, \mathcal{A}, P, R, \gamma)$:

- $\mathcal{S}$: **State space** (finite or infinite set of states)
- $\mathcal{A}$: **Action space** (finite or infinite set of actions)
- $P(s'|s,a)$: **Transition probability** - probability of reaching $s'$ from $s$ by taking $a$
- $R(s,a,s')$: **Reward function** - immediate reward for transition
- $\gamma \in [0,1)$: **Discount factor** - importance of future rewards

### 10.2.2 Markov Property

**Future is independent of past given present**:

$$P(s_{t+1}|s_t, a_t, s_{t-1}, a_{t-1}, ..., s_0, a_0) = P(s_{t+1}|s_t, a_t)$$

Current state captures all relevant information.

### 10.2.3 Policy

**Policy** $\pi$: Mapping from states to actions

**Deterministic**: $a = \pi(s)$

**Stochastic**: $\pi(a|s) = P(a_t = a | s_t = s)$

### 10.2.4 Return

**Return** $G_t$: Cumulative discounted reward from time $t$

$$G_t = r_t + \gamma r_{t+1} + \gamma^2 r_{t+2} + ... = \sum_{k=0}^{\infty} \gamma^k r_{t+k}$$

**Why discount?**
1. Mathematical convenience (ensures convergence)
2. Uncertainty about future
3. Preference for immediate rewards
4. Avoids infinite returns in continuing tasks

**Special cases**:
- $\gamma = 0$: Only immediate reward matters (myopic)
- $\gamma \to 1$: Far-sighted (all rewards equally important)

### 10.2.5 Value Functions

**State-Value Function** $V^\pi(s)$: Expected return starting from $s$, following $\pi$

$$V^\pi(s) = \mathbb{E}_\pi[G_t | s_t = s] = \mathbb{E}_\pi\left[\sum_{k=0}^{\infty} \gamma^k r_{t+k} \mid s_t = s\right]$$

**Action-Value Function** $Q^\pi(s,a)$: Expected return starting from $s$, taking $a$, then following $\pi$

$$Q^\pi(s,a) = \mathbb{E}_\pi[G_t | s_t = s, a_t = a] = \mathbb{E}_\pi\left[\sum_{k=0}^{\infty} \gamma^k r_{t+k} \mid s_t = s, a_t = a\right]$$

**Relationship**:
$$V^\pi(s) = \sum_{a} \pi(a|s) Q^\pi(s,a)$$

## 10.3 Bellman Equations

**Bellman equations** express recursive relationship between value of state and its successors.

### 10.3.1 Bellman Expectation Equations

**For state-value**:
$$\begin{align}
V^\pi(s) &= \mathbb{E}_\pi[r_t + \gamma V^\pi(s_{t+1}) | s_t = s] \\
&= \sum_a \pi(a|s) \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V^\pi(s')]
\end{align}$$

**For action-value**:
$$\begin{align}
Q^\pi(s,a) &= \mathbb{E}_\pi[r_t + \gamma Q^\pi(s_{t+1}, a_{t+1}) | s_t = s, a_t = a] \\
&= \sum_{s'} P(s'|s,a)\left[R(s,a,s') + \gamma \sum_{a'} \pi(a'|s') Q^\pi(s',a')\right]
\end{align}$$

### 10.3.2 Bellman Optimality Equations

**Optimal value functions**:
$$V^*(s) = \max_\pi V^\pi(s)$$
$$Q^*(s,a) = \max_\pi Q^\pi(s,a)$$

**Bellman optimality for $V^*$**:
$$V^*(s) = \max_a \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V^*(s')]$$

**Bellman optimality for $Q^*$**:
$$Q^*(s,a) = \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma \max_{a'} Q^*(s',a')]$$

**Optimal policy**:
$$\pi^*(s) = \arg\max_a Q^*(s,a)$$

### 10.3.3 Key Insights

1. **Recursive structure**: Value of state equals immediate reward + discounted value of next state
2. **Expectation over outcomes**: Must consider all possible next states
3. **Max for optimality**: Optimal value takes best action
4. **Fixed point**: Bellman equations define unique solution

## 10.4 Dynamic Programming

**Dynamic Programming (DP)** solves MDPs when model is known (transition probabilities and rewards).

### 10.4.1 Policy Evaluation

**Problem**: Compute $V^\pi$ for given policy $\pi$

**Iterative update**: Use Bellman expectation equation

$$V_{k+1}(s) = \sum_a \pi(a|s) \sum_{s'} P(s'|s,a)[R(s,a,s') + \gamma V_k(s')]$$

Converges to $V^\pi$ as $k \to \infty$

### 10.4.2 Policy Iteration

**Idea**: Alternate between evaluating policy and improving it

```
Algorithm: POLICY-ITERATION(MDP)

begin
    // Initialize
π ← arbitrary policy
    V ← arbitrary values
    
    repeat
        // Policy Evaluation: Compute V^π
        repeat
            Δ ← 0
            for each state s do
                v ← V[s]
                V[s] ← ∑_a π(a|s) ∑_{s'} P(s'|s,a)[R(s,a,s') + γ V[s']]
Δ ← max(Δ, |v - V[s]|)
        until Δ < θ  // Convergence threshold
        
        // Policy Improvement
        policy_stable ← True
        for each state s do
            old_action ← π(s)
            π(s) ← argmax_a ∑_{s'} P(s'|s,a)[R(s,a,s') + γ V[s']]
            if old_action ≠ π(s) then
                policy_stable ← False
        
    until policy_stable
    
    return π, V
end
```

**Convergence**: Guaranteed to converge to $\pi^*$ in finite steps

**Complexity**: $O(|\mathcal{S}|^2|\mathcal{A}|)$ per iteration

### 10.4.3 Value Iteration

**Idea**: Combine evaluation and improvement in single update

```
Algorithm: VALUE-ITERATION(MDP, threshold θ)

begin
    // Initialize
    V ← arbitrary values (often V[s] = 0 for all s)
    
    repeat
Δ ← 0
        
        // Update all states
        for each state s do
            v ← V[s]
            
            // Bellman optimality backup
            V[s] ← max_a ∑_{s'} P(s'|s,a)[R(s,a,s') + γ V[s']]
            
Δ ← max(Δ, |v - V[s]|)
    
    until Δ < θ
    
    // Extract optimal policy
    for each state s do
π(s) ← argmax_a ∑_{s'} P(s'|s,a)[R(s,a,s') + γ V[s']]
    
    return π, V
end
```

**Convergence**: Guaranteed to converge to $V^*$

**Complexity**: $O(|\mathcal{S}|^2|\mathcal{A}|)$ per iteration

**Faster than Policy Iteration**: No need for complete policy evaluation

### 10.4.4 DP Limitations

1. **Requires model**: Must know $P(s'|s,a)$ and $R(s,a,s')$
2. **Computational cost**: $O(|\mathcal{S}|^2|\mathcal{A}|)$ per iteration
3. **Curse of dimensionality**: Exponential growth with state variables

**Solution**: Model-free methods (Monte Carlo, TD learning)

## 10.5 Monte Carlo Methods

**Monte Carlo (MC)** learns from complete episodes without requiring model.

### 10.5.1 Key Idea

**Sample returns** from actual experience:
1. Generate episode following policy $\pi$
2. For each state $s$ visited, observe actual return $G_t$
3. Average returns to estimate $V^\pi(s)$

**Unbiased estimate**: $\mathbb{E}[G_t | s_t = s] = V^\pi(s)$

### 10.5.2 First-Visit MC

```
Algorithm: FIRST-VISIT-MC(policy π, num_episodes)

begin
    V ← arbitrary values
    Returns[s] ← empty list for all s
    
    for episode = 1 to num_episodes do
        // Generate episode following π
        S_0, A_0, R_1, S_1, A_1, R_2, ..., S_T → GENERATE-EPISODE(π)
        
        G ← 0
        
        // Process episode backwards
        for t = T-1 down to 0 do
            G ← γ G + R_{t+1}
            
            // If first visit to S_t in this episode
            if S_t not in {S_0, S_1, ..., S_{t-1}} then
                Returns[S_t].append(G)
                V[S_t] ← average(Returns[S_t])
    
    return V
end
```

**Every-Visit MC**: Update for every visit (not just first)

### 10.5.3 MC for Action-Values

**Problem**: Need to estimate $Q^\pi(s,a)$ for policy improvement

**Challenge**: Some $(s,a)$ pairs may never be visited

**Solution**: Exploring starts - ensure all pairs have nonzero probability

### 10.5.4 MC Control

```
Algorithm: MC-CONTROL(num_episodes)

begin
    Q ← arbitrary values
π ← arbitrary policy
    Returns[s,a] ← empty list for all s,a
    
    for episode = 1 to num_episodes do
        // Generate episode with exploring starts
        S_0, A_0 → random state-action pair
        S_0, A_0, R_1, ..., S_T → GENERATE-EPISODE(π)
        
        G ← 0
        for t = T-1 down to 0 do
            G ← γ G + R_{t+1}
            
            if (S_t, A_t) not seen earlier in episode then
                Returns[S_t, A_t].append(G)
                Q[S_t, A_t] ← average(Returns[S_t, A_t])
                
                // Policy improvement
π(S_t) ← argmax_a Q[S_t, a]
    
    return π, Q
end
```

### 10.5.5 MC Properties

**Advantages**:
- Model-free: No need for $P$ or $R$
- Can learn from actual or simulated experience
- Simple to understand and implement
- Unbiased estimates

**Disadvantages**:
- Requires episodic tasks (must terminate)
- High variance (depends on entire episode)
- Must wait until end of episode to update
- Slow learning

**Complexity**: $O(|\mathcal{S}||\mathcal{A}|)$ per episode

## 10.6 Temporal Difference Learning

**Temporal Difference (TD)** combines ideas from DP and MC - learns from incomplete episodes using bootstrapping.

### 10.6.1 TD(0) Prediction

**Key idea**: Update value estimate using next state's estimate

**TD update**:
$$V(s_t) \leftarrow V(s_t) + \alpha[r_{t+1} + \gamma V(s_{t+1}) - V(s_t)]$$

**TD error** $\delta_t$:
$$\delta_t = r_{t+1} + \gamma V(s_{t+1}) - V(s_t)$$

```
Algorithm: TD(0)-PREDICTION(policy π, learning_rate α)

begin
    V ← arbitrary values (e.g., V[s] = 0 for all s)
    
    for each episode do
        s ← initial state
        
        for each step in episode do
            a → action from π(s)
            r, s' → take action a, observe reward and next state
            
            // TD update
            V[s] ← V[s] + α[r + γ V[s'] - V[s]]
            
            s ← s'
            if s is terminal then break
    
    return V
end
```

### 10.6.2 TD vs MC vs DP

| Property | DP | MC | TD |
|----------|----|----|---|
| Model-free | No | Yes | Yes |
| Bootstraps | Yes | No | Yes |
| Episodic required | No | Yes | No |
| Bias | Zero | Zero | Biased |
| Variance | Low | High | Medium |
| Convergence speed | Fast | Slow | Medium |

**Bootstrapping**: Update estimate based on other estimates

### 10.6.3 SARSA (On-Policy TD Control)

**SARSA**: State-Action-Reward-State-Action

**Update rule**:
$$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha[r_{t+1} + \gamma Q(s_{t+1}, a_{t+1}) - Q(s_t, a_t)]$$

```
Algorithm: SARSA(learning_rate α, discount γ, exploration ε)

begin
    Q ← arbitrary values
    
    for each episode do
        s ← initial state
        a → action from ε-greedy policy based on Q
        
        for each step in episode do
            r, s' → take action a
            a' → action from ε-greedy policy for s'
            
            // SARSA update
            Q[s,a] ← Q[s,a] + α[r + γ Q[s',a'] - Q[s,a]]
            
            s ← s'
            a ← a'
            if s is terminal then break
    
    return Q
end
```

**ε-greedy policy**:
$$\pi(a|s) = \begin{cases} 
1-\epsilon + \epsilon/|\mathcal{A}| & \text{if } a = \arg\max_a Q(s,a) \\
\epsilon/|\mathcal{A}| & \text{otherwise}
\end{cases}$$

**On-policy**: Learns about policy being followed

### 10.6.4 Q-Learning (Off-Policy TD Control)

**Key difference**: Use max over next actions instead of following policy

**Update rule**:
$$Q(s_t, a_t) \leftarrow Q(s_t, a_t) + \alpha[r_{t+1} + \gamma \max_{a'} Q(s_{t+1}, a') - Q(s_t, a_t)]$$

```
Algorithm: Q-LEARNING(α, γ, ε)

begin
    Q ← arbitrary values
    
    for each episode do
        s ← initial state
        
        for each step in episode do
            // Choose action (can be ε-greedy)
            a → action from ε-greedy policy based on Q
            
            r, s' → take action a
            
            // Q-learning update (uses max)
            Q[s,a] ← Q[s,a] + α[r + γ max_a' Q[s',a'] - Q[s,a]]
            
            s ← s'
            if s is terminal then break
    
    return Q
end
```

**Off-policy**: Learns optimal policy while following exploratory policy

**Convergence**: Proven to converge to $Q^*$ under mild conditions

### 10.6.5 SARSA vs Q-Learning

| Aspect | SARSA | Q-Learning |
|--------|-------|------------|
| Type | On-policy | Off-policy |
| Update | Uses next action taken | Uses max action |
| Policy learned | Policy being followed | Optimal policy |
| Exploration impact | Affects learning | Doesn't affect target |
| Cliff walking | Safe path | Optimal path |
| Convergence | To policy being followed | To optimal policy |

## 10.7 Deep Q-Networks (DQN)

**DQN** uses deep neural networks to approximate Q-function, enabling RL in large state spaces.

### 10.7.1 Function Approximation

**Problem**: Tabular methods don't scale to large/continuous state spaces

**Solution**: Approximate Q-function with neural network

$$Q(s,a) \approx Q(s,a;\boldsymbol{\theta})$$

where $\boldsymbol{\theta}$ are network parameters.

### 10.7.2 Challenges

**1. Correlated samples**: Sequential states are highly correlated
**2. Non-stationary targets**: Target keeps changing as Q is updated
**3. Divergence**: Function approximation can be unstable

### 10.7.3 DQN Solutions

**Experience Replay**:
- Store transitions $(s,a,r,s')$ in replay buffer
- Sample random minibatches for training
- Breaks correlations, improves data efficiency

**Target Network**:
- Use separate network $Q(s,a;\boldsymbol{\theta}^-)$ for computing targets
- Update $\boldsymbol{\theta}^-$ periodically (e.g., every 10k steps)
- Stabilizes training

### 10.7.4 DQN Algorithm

```
Algorithm: DQN(env, network, buffer_size, batch_size)

begin
    // Initialize
    Q ← neural network with random weights θ
    Q_target ← copy of Q with weights θ⁻
    D ← replay buffer with capacity buffer_size
    
    for episode = 1 to num_episodes do
        s ← initial state
        
        for t = 1 to max_steps do
            // Select action (ε-greedy)
            with probability ε:
                a → random action
            otherwise:
                a → argmax_a' Q(s, a'; θ)
            
            // Execute action
            r, s' → env.step(a)
            
            // Store transition
            D.store((s, a, r, s'))
            
            // Sample minibatch
            B → sample batch_size transitions from D
            
            // Compute targets
            for each (s_i, a_i, r_i, s'_i) in B do
                if s'_i is terminal:
                    y_i ← r_i
                else:
                    y_i ← r_i + γ max_a' Q_target(s'_i, a'; θ⁻)
            
            // Gradient descent step
            loss ← (1/batch_size) ∑_i (y_i - Q(s_i, a_i; θ))²
θ ← θ - α ∇_θ loss
            
            // Update target network periodically
            if t mod C == 0 then
θ⁻ ← θ
            
            s ← s'
            if s is terminal then break
    
    return Q
end
```

### 10.7.5 DQN Improvements

**Double DQN**: Addresses overestimation bias
$$y = r + \gamma Q(s', \arg\max_{a'} Q(s',a';\boldsymbol{\theta}); \boldsymbol{\theta}^-)$$

**Dueling DQN**: Separate value and advantage streams
$$Q(s,a) = V(s) + A(s,a) - \frac{1}{|\mathcal{A}|}\sum_{a'} A(s,a')$$

**Prioritized Experience Replay**: Sample important transitions more frequently

## 10.8 Policy Gradient Methods

**Policy Gradient** directly parameterizes and optimizes policy $\pi(a|s;\boldsymbol{\theta})$.

### 10.8.1 Motivation

**Advantages over value-based methods**:
1. Can learn stochastic policies
2. Better convergence properties
3. Effective in high-dimensional/continuous action spaces
4. Can incorporate domain knowledge

### 10.8.2 Objective

**Maximize expected return**:
$$J(\boldsymbol{\theta}) = \mathbb{E}_{\tau \sim \pi_\theta}[R(\tau)]$$

where $\tau = (s_0, a_0, r_1, s_1, a_1, ...)$ is a trajectory.

### 10.8.3 Policy Gradient Theorem

**Gradient of objective**:
$$\nabla_\boldsymbol{\theta} J(\boldsymbol{\theta}) = \mathbb{E}_{\tau \sim \pi_\theta}\left[\sum_{t=0}^{T} \nabla_\boldsymbol{\theta} \log \pi(a_t|s_t;\boldsymbol{\theta}) G_t\right]$$

**Key insight**: Don't need to differentiate environment dynamics

### 10.8.4 REINFORCE Algorithm

```
Algorithm: REINFORCE(policy network π_θ, learning_rate α)

begin
    Initialize policy parameters θ randomly
    
    for episode = 1 to num_episodes do
        // Generate episode
        τ ← {s_0, a_0, r_1, ..., s_T} following π_θ
        
        // Compute returns
        for t = 0 to T-1 do
            G_t ← ∑_{k=t}^{T-1} γ^{k-t} r_{k+1}
        
        // Update policy
        for t = 0 to T-1 do
θ ← θ + α γ^t G_t ∇_θ log π_θ(a_t|s_t)
    
    return θ
end
```

**Problem**: High variance (entire episode return used)

### 10.8.5 Baseline Reduction

**Subtract baseline** to reduce variance without introducing bias:
$$\nabla_\boldsymbol{\theta} J(\boldsymbol{\theta}) = \mathbb{E}\left[\sum_{t=0}^{T} \nabla_\boldsymbol{\theta} \log \pi(a_t|s_t;\boldsymbol{\theta}) (G_t - b(s_t))\right]$$

**Common choice**: $b(s) = V(s)$ (state value)

**Advantage**: $A(s,a) = Q(s,a) - V(s)$
- Measures how much better action $a$ is than average

### 10.8.6 Actor-Critic

**Combine policy gradient (actor) with value function (critic)**

**Actor**: Policy $\pi(a|s;\boldsymbol{\theta})$
**Critic**: Value function $V(s;\mathbf{w})$

```
Algorithm: ACTOR-CRITIC(α_θ, α_w, γ)

begin
    Initialize actor parameters θ, critic parameters w
    
    for each episode do
        s ← initial state
        I ← 1  // discount factor
        
        for each step do
            a ~ π(a|s; θ)
            r, s' → take action a
            
            // TD error (advantage estimate)
δ ← r + γ V(s'; w) - V(s; w)
            
            // Update critic
            w ← w + α_w δ ∇_w V(s; w)
            
            // Update actor
θ ← θ + α_θ I δ ∇_θ log π(a|s; θ)
            
            I ← γ I
            s ← s'
            if s is terminal then break
    
    return θ, w
end
```

**Advantages**:
- Lower variance than REINFORCE
- Can learn online (no need to wait for episode end)
- Works for continuing tasks

### 10.8.7 Advanced: A3C, PPO, TRPO

**A3C** (Asynchronous Advantage Actor-Critic):
- Multiple parallel actors
- Stabilizes training

**PPO** (Proximal Policy Optimization):
- Constrains policy updates
- Clip objective: $\min(r_t(\boldsymbol{\theta})A_t, \text{clip}(r_t(\boldsymbol{\theta}), 1-\epsilon, 1+\epsilon)A_t)$
- State-of-the-art for many tasks

**TRPO** (Trust Region Policy Optimization):
- Guarantees monotonic improvement
- Constrains KL divergence between policies

### Key Concepts

**MDPs**:
- States, actions, transitions, rewards
- Markov property
- Bellman equations

**Value Functions**:
- $V^\pi(s)$: State value
- $Q^\pi(s,a)$: Action value
- $V^*(s)$, $Q^*(s,a)$: Optimal values

**Methods**:
- **DP**: Model-based, exact
- **MC**: Model-free, high variance, unbiased
- **TD**: Model-free, low variance, biased
- **Deep RL**: Function approximation, scales
- **Policy Gradient**: Direct policy optimization

### Exploration-Exploitation

**$\epsilon$-greedy**: Explore with probability $\epsilon$
**Softmax**: $P(a) \propto e^{Q(s,a)/\tau}$
**UCB**: $a = \arg\max_a Q(s,a) + c\sqrt{\frac{\ln t}{N(s,a)}}$

### Practical Tips

1. **Start simple**: Tabular methods for small problems
2. **Normalize rewards**: Makes learning stable
3. **Tune learning rate**: Critical for convergence
4. **Use baselines**: Reduces variance in policy gradients
5. **Monitor training**: Plot returns, Q-values
6. **Debug thoroughly**: RL is notoriously difficult
