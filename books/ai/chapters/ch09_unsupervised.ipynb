```
Algorithm: K-MEANS(data, k)

begin
    centroids ← randomly select k points
    
    repeat
        // Assignment step
        for each point x do
            assign x to nearest centroid
        
        // Update step
        for each cluster c do
            centroid[c] ← mean of points in c
    
    until centroids stop changing
    
    return centroids, assignments
end
```

## 9.2 PCA

**Principal Component Analysis:**

$$X_{\text{reduced}} = X \cdot W_k$$

Where $W_k$ contains top k eigenvectors of covariance matrix.

## 9.3 Implementation

See: [ch09_unsupervised_implementation.ipynb](ch09_unsupervised_implementation.ipynb)

Includes:
- k-Means clustering
- Hierarchical clustering
- PCA algorithm
- Autoencoders
- Real-world: Customer segmentation
