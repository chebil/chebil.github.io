## Exercise 5: Real Dataset Comparison

Compare algorithms on real datasets.

````python
from sklearn.datasets import load_iris, load_breast_cancer, load_wine
from sklearn.preprocessing import StandardScaler
import pandas as pd

def evaluate_models(X, y, dataset_name):
    """Evaluate all models on dataset"""
    # Standardize features
    scaler = StandardScaler()
    X_scaled = scaler.fit_transform(X)
    
    results = {}
    for name, (model_class, params) in models.items():
        scores, mean, std = cross_validate(model_class, X_scaled, y, k=5, **params)
        results[name] = {'mean': mean, 'std': std, 'scores': scores}
    
    return results

# Load datasets
datasets = {
    'Iris': load_iris(return_X_y=True),
    'Breast Cancer': load_breast_cancer(return_X_y=True),
    'Wine': load_wine(return_X_y=True)
}

# Evaluate all
all_results = {}
for ds_name, (X, y) in datasets.items():
    print(f'\nDataset: {ds_name}')
    print(f'Samples: {X.shape[0]}, Features: {X.shape[1]}')
    results = evaluate_models(X, y, ds_name)
    all_results[ds_name] = results
    
    for model_name, res in results.items():
        print(f'{model_name:25s}: {res["mean"]:.3f} (+/- {res["std"]:.3f})')

# Visualize comparison
df_results = pd.DataFrame({
    ds: {model: results[model]['mean'] for model in results}
    for ds, results in all_results.items()
})

df_results.plot(kind='bar', figsize=(10, 6))
plt.xlabel('Model')
plt.ylabel('Accuracy')
plt.title('Model Comparison Across Datasets')
plt.legend(title='Dataset')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````
