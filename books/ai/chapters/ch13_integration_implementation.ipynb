```python
import numpy as np
from typing import Dict, List, Any

print('Imports complete')
```

```python
class PerceptionModule:
    """Neural network for feature extraction"""
    def __init__(self):
        self.model = None
    
    def process(self, raw_input):
        # Simulate neural feature extraction
        features = {
            'entities': self._extract_entities(raw_input),
            'intent': self._classify_intent(raw_input),
            'confidence': 0.85
        }
        return features
    
    def _extract_entities(self, text):
        # Simplified NER
        return ['entity1', 'entity2']
    
    def _classify_intent(self, text):
        # Simplified intent classification
        return 'query'

print('Perception module defined')
```

```python
class ReasoningModule:
    """Symbolic reasoning engine"""
    def __init__(self):
        self.kb = {}
        self.rules = []
    
    def add_fact(self, fact):
        self.kb[fact] = True
    
    def add_rule(self, premises, conclusion):
        self.rules.append((premises, conclusion))
    
    def reason(self, query, features):
        # Combine neural features with symbolic reasoning
        result = {
            'answer': None,
            'justification': [],
            'confidence': features.get('confidence', 0.5)
        }
        
        # Simple rule matching
        for premises, conclusion in self.rules:
            if all(p in self.kb for p in premises):
                result['answer'] = conclusion
                result['justification'] = premises
                break
        
        return result

print('Reasoning module defined')
```

```python
class HybridAISystem:
    """Integrated neuro-symbolic AI system"""
    def __init__(self):
        self.perception = PerceptionModule()
        self.reasoning = ReasoningModule()
    
    def process_query(self, query):
        print(f'Query: {query}')
        
        # Step 1: Neural perception
        features = self.perception.process(query)
        print(f'Features extracted: {features}')
        
        # Step 2: Symbolic reasoning
        result = self.reasoning.reason(query, features)
        print(f'Reasoning result: {result}')
        
        # Step 3: Generate response
        response = self._generate_response(result)
        return response
    
    def _generate_response(self, result):
        if result['answer']:
            return {
                'answer': result['answer'],
                'explanation': f"Based on: {result['justification']}",
                'confidence': result['confidence']
            }
        return {'answer': 'Unknown', 'confidence': 0.0}

print('Hybrid AI system defined')
```

```python
print('=== Hybrid AI System Demo ===')

# Create system
system = HybridAISystem()

# Add knowledge
system.reasoning.add_fact('hasFur')
system.reasoning.add_fact('produceMilk')
system.reasoning.add_rule(['hasFur', 'produceMilk'], 'mammal')

# Process query
response = system.process_query('What kind of animal is it?')
print(f'\nFinal Response: {response}')
```
