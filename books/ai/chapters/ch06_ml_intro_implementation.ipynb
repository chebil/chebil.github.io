```python
import numpy as np
import math
from collections import Counter

print('Imports complete')
```

```python
def entropy(labels):
    counts = Counter(labels)
    probs = [count/len(labels) for count in counts.values()]
    return -sum(p * math.log2(p) for p in probs if p > 0)

def information_gain(data, labels, attr_idx):
    total_entropy = entropy(labels)
    values = set(row[attr_idx] for row in data)
    weighted_entropy = 0
    
    for v in values:
        subset_labels = [labels[i] for i, row in enumerate(data) if row[attr_idx] == v]
        weighted_entropy += (len(subset_labels) / len(labels)) * entropy(subset_labels)
    
    return total_entropy - weighted_entropy

print('Entropy and gain functions defined')
```

```python
class DecisionTree:
    def __init__(self, max_depth=10):
        self.max_depth = max_depth
        self.tree = None
    
    def fit(self, X, y, depth=0):
        if len(set(y)) == 1:
            return y[0]
        if depth >= self.max_depth or len(X[0]) == 0:
            return Counter(y).most_common(1)[0][0]
        
        gains = [information_gain(X, y, i) for i in range(len(X[0]))]
        best_attr = gains.index(max(gains))
        
        tree = {"attr": best_attr, "branches": {}}
        for value in set(row[best_attr] for row in X):
            subset_X = [row for row in X if row[best_attr] == value]
            subset_y = [y[i] for i, row in enumerate(X) if row[best_attr] == value]
            tree["branches"][value] = self.fit(subset_X, subset_y, depth + 1)
        
        return tree
    
    def train(self, X, y):
        self.tree = self.fit(X, y)

print('Decision Tree implemented')
```
