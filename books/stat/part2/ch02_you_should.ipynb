## 2.4.1 Remember These Definitions

**Correlation coefficient** — A number, usually written $r$, that measures the strength and direction of the *linear* relationship between two variables. For standardized data $\hat{x}_i$ and $\hat{y}_i$,

$$r = \frac{1}{N} \sum_{i=1}^{N} \hat{x}_i \, \hat{y}_i, \qquad \hat{x}_i = \frac{x_i - \bar{x}}{\sigma_x}, \quad \hat{y}_i = \frac{y_i - \bar{y}}{\sigma_y}.$$

**Covariance** — The average product of deviations from the means, $\text{cov}(x, y) = \frac{1}{N}\sum_{i=1}^{N}(x_i - \bar{x})(y_i - \bar{y})$. Correlation is the covariance of the *standardized* variables.

**Scatter plot** — A plot of paired observations $(x_i, y_i)$ as points, used to reveal the shape, direction, and strength of a relationship.

## 2.4.3 Remember These Facts

- $-1 \le r \le 1$. Values near $\pm 1$ indicate a strong linear trend; values near $0$ indicate little *linear* association.
- The sign of $r$ gives the direction: positive means $y$ tends to increase with $x$; negative means $y$ tends to decrease.
- $r$ is **symmetric** ($r_{xy} = r_{yx}$) and **unitless**; it is unchanged by shifting or rescaling either variable.
- $r$ measures **linear** association only. A strong nonlinear relationship (e.g., a parabola) can have $r \approx 0$.
- $r$ is **sensitive to outliers**: a single extreme point can inflate or hide a correlation.
- **Correlation is not causation.** A lurking variable can produce a strong correlation between two variables that do not influence each other.

## 2.4.4 Use These Procedures

**To compute and interpret a correlation:**
1. Plot the data as a scatter plot first — check for a roughly linear trend and outliers.
2. Standardize both variables (subtract the mean, divide by the standard deviation).
3. Compute $r$ as the average product of the standardized values (e.g., `np.corrcoef(x, y)`).
4. Report both the **sign** (direction) and **magnitude** (strength), and note any outliers.

**To predict $y$ from $x$ using the best linear predictor** (in standardized coordinates):

$$\hat{y}^p = r \, \hat{x}, \qquad \text{with typical error (RMSE)} = \sqrt{1 - r^2}.$$
